```python
def create_loss_curve(filename, training_type='finetune'):
    

    df = pd.read_csv('data/' + filename + '.csv')
    
    if training_type=='finetune':
        pattern = r"{'loss': ([\d.]+), 'grad_norm': ([\d.]+), 'learning_rate': ([\de.-]+), 'epoch': ([\d.]+)}"

        loss_list = []
        grad_norm_list = []
        learning_rate_list = []
        epoch_list = []

        for index, row in df.iterrows():
            # Search for the pattern in the text
            match = re.search(pattern, row['message'])
            if match:
                # Extract values and append them to respective lists
                loss_list.append(float(match.group(1)))
                grad_norm_list.append(float(match.group(2)))
                learning_rate_list.append(float(match.group(3)))
                epoch_list.append(float(match.group(4)))

        extracted_df = pd.DataFrame({
            'loss': loss_list,
            'grad_norm': grad_norm_list,
            'learning_rate': learning_rate_list,
            'epoch': epoch_list})
        
        
    elif training_type =='pretrain':
        pattern = r"{'loss': ([\d.]+), 'learning_rate': ([\de.-]+), 'epoch': ([\d.]+)}"

        loss_list = []
        learning_rate_list = []
        epoch_list = []

        for index, row in df.iterrows():
            # Search for the pattern in the text
            match = re.search(pattern, row['message'])
            if match:
                # Extract values and append them to respective lists
                loss_list.append(float(match.group(1)))
                learning_rate_list.append(float(match.group(2)))
                epoch_list.append(float(match.group(3)))

        extracted_df = pd.DataFrame({
            'loss': loss_list,
            'learning_rate': learning_rate_list,
            'epoch': epoch_list
        })

    
    plt.plot(extracted_df['epoch'], extracted_df['loss'], marker='o', linestyle='-')
    plt.title(f'Loss Curve for {filename}')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.grid(True)
    plt.show()
    
    print(extracted_df)
    return extracted_df
```

```python
def create_eval_loss_curve(filename):
    df = pd.read_csv('data/' + filename + '.csv')
    # pattern = r"{'eval_loss': ([\d.]+), 'eval_rougeL': ([\d.]+), 'eval_runtime': ([\d.]+.[\d]+), 'eval_samples_per_second': ([\d.]+), 'eval_steps_per_second': ([\d.]+), 'epoch': ([\d.]+)}"
    pattern = r"{'eval_loss': ([\d.]+), 'eval_runtime': ([\d.]+.[\d]+), 'eval_samples_per_second': ([\d.]+), 'eval_steps_per_second': ([\d.]+), 'epoch': ([\d.]+)}"

    eval_loss_list = []
    # eval_rougeL_list = []
    eval_runtime_list = []
    eval_samples_per_second_list = []
    eval_steps_per_second_list = []
    epoch_list = []

    for index, row in df.iterrows():
        match = re.search(pattern, row['message'])
        if match:
            # Extract values and append them to respective lists
            eval_loss_list.append(float(match.group(1)))
            # eval_rougeL_list.append(float(match.group(2)))
            eval_runtime_list.append(float(match.group(2)))
            eval_samples_per_second_list.append(float(match.group(3)))
            eval_steps_per_second_list.append(float(match.group(4)))
            epoch_list.append(float(match.group(5)))

    # Create a new DataFrame with extracted values
    extracted_df = pd.DataFrame({
        'eval_loss': eval_loss_list,
        # 'eval_rougeL': eval_rougeL_list,
        'eval_runtime': eval_runtime_list,
        'eval_samples_per_second': eval_samples_per_second_list,
        'eval_steps_per_second': eval_steps_per_second_list,
        'epoch': epoch_list
    })

    # Plotting the loss curve
    plt.plot(extracted_df['epoch'], extracted_df['eval_loss'], marker='o', linestyle='-')
    plt.title(f'Evaluation Loss Curve for {filename}')
    plt.xlabel('Epoch')
    plt.ylabel('Evaluation Loss')
    plt.grid(True)
    plt.show()
    
    print(extracted_df)
    return extracted_df
    
```
